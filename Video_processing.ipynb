```python
import yt_dlp
import cv2
import os

# Use the correct file path for the downloaded video and the correct extension
vid = cv2.VideoCapture("/content/Ride with Dark Fury ｜｜ Crazy Rider [hNRyg7Ci0S0].webm")

output_folder = "raw_videos"
os.makedirs(output_folder, exist_ok=True)
studentName = "oishi"
output_path = os.path.join(output_folder, f"yt_bikeride_{studentName}.mp4")
codec = cv2.VideoWriter_fourcc(*'mp4v')
# fps = vid.get(cv2.CAP_PROP_FPS)
width = int(vid.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(vid.get(cv2.CAP_PROP_FRAME_HEIGHT))
out = cv2.VideoWriter(output_path, codec, 30, (width, height))

while True:
    # Read from the cv2.VideoCapture object 'vid'
    success, frame = vid.read()
    if not success:
        break
    out.write(frame)
    # cv2.imshow("Video", frame) # Removed as it's not supported in Colab
    # if cv2.waitKey(1) & 0xFF == ord('q'): # Removed as it's not supported in Colab
    #     break

vid.release()
out.release()
# cv2.destroyAllWindows() # Removed as it's not supported in Colab
```

```python
import os

FRAMES_TO_SAVE = 20
FRAME_INTERVAL = 30

# yt = YouTube(video_url) # Removed as yt-dlp was used for download
os.makedirs("input", exist_ok=True)
os.makedirs("output", exist_ok=True)
```

```python
import cv2

# Use cv2 to open the downloaded video file
video_path = "/content/Ride with Dark Fury ｜｜ Crazy Rider [hNRyg7Ci0S0].webm" # Replace with the actual path to your downloaded video
vidcap = cv2.VideoCapture(video_path)

success,image = vidcap.read()
count = 0
saved_frames = 0

while success and saved_frames < FRAMES_TO_SAVE:
  if count % FRAME_INTERVAL == 0:
    cv2.imwrite(os.path.join("input", f"frame{count:04d}.jpg"), image)     # save frame as JPEG file
    saved_frames += 1
  success,image = vidcap.read()
  count += 1
```

```python
cap = cv2.VideoCapture(video_path)
frame_count = 0
saved_frames = 0

while True:
    ret, frame = cap.read()
    if not ret or saved_frames >= FRAMES_TO_SAVE:
        break

    # Save frame every FRAME_INTERVAL frames
    if frame_count % FRAME_INTERVAL == 0:
        frame_filename = os.path.join("output", f"frame_{saved_frames}.jpg")
        cv2.imwrite(frame_filename, frame)
        print(f"Saved {frame_filename}")
        saved_frames += 1

    frame_count += 1

cap.release()
cv2.destroyAllWindows()
```
